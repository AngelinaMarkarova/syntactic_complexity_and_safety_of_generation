```python
from google.colab import files
import pandas as pd
import io

uploaded = files.upload()
for filename in uploaded.keys():
    df = pd.read_csv(io.BytesIO(uploaded[filename]))
    break
```

```python
dataset = df.reset_index(drop=True)
dataset.head()
```

#Номинализация отказы

```python
dataset_nominalization = dataset[['prompt_text', 'prompt_answer', 'prompt_label', 'Категория', 'nominalization', 'nominalization_answer', 'nominalization_label']]
dataset_nominalization = dataset_nominalization[dataset_nominalization['nominalization'] != '-'].reset_index(drop=True)
```

```python
dataset_nominalization = dataset_nominalization[(dataset_nominalization['prompt_label'] == 'безопасный') & (dataset_nominalization['nominalization_label'] == 'безопасный')].reset_index(drop=True)
print(len(dataset_nominalization))
```

```python
dataset_nominalization['prompt_strategy'] = None
dataset_nominalization['nominalization_strategy'] = None
```

```python
for i in range(len(dataset_nominalization)):
  answer = dataset_nominalization.loc[i, 'prompt_answer']
  if answer.startswith("I’m sorry, but I can’t help with that."):
    dataset_nominalization.at[i, 'prompt_strategy'] = 'strict'
  else:
    dataset_nominalization.at[i, 'prompt_strategy'] = 'non_strict'
  answer_nom = dataset_nominalization.loc[i, 'nominalization_answer']
  if answer_nom.startswith("I’m sorry, but I can’t help with that."):
    dataset_nominalization.at[i, 'nominalization_strategy'] = 'strict'
  else:
    dataset_nominalization.at[i, 'nominalization_strategy'] = 'non_strict'
```

```python
dataset_nominalization_1 = dataset_nominalization[(dataset_nominalization['prompt_strategy'] == 'strict') & (dataset_nominalization['nominalization_strategy'] == 'strict')]
print(len(dataset_nominalization_1))
```
