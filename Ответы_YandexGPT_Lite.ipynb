```python
import openai

YANDEX_CLOUD_FOLDER = "_"
YANDEX_CLOUD_API_KEY = "_"
YANDEX_CLOUD_MODEL = "yandexgpt-5-lite/latest"

client = openai.OpenAI(
  api_key=YANDEX_CLOUD_API_KEY,
  base_url="https://ai.api.cloud.yandex.net/v1",
  project=YANDEX_CLOUD_FOLDER
)
```

#Загрузка датасета

```python
from google.colab import files
import pandas as pd
import io

uploaded = files.upload()
for filename in uploaded.keys():
    df = pd.read_csv(io.BytesIO(uploaded[filename]))
    break
```

```python
dataset = df.reset_index(drop=True)
dataset.head()
```

#Получение ответов

```python
def yandex(question):
  response = client.chat.completions.create(
    model=f"gpt://{YANDEX_CLOUD_FOLDER}/{YANDEX_CLOUD_MODEL}",
    messages=[
        {"role": "user", "content": question}
    ],
    max_tokens=2000,
    temperature=0.5
  )

  return response.choices[0].message.content
```

```python
for i in range(3, len(dataset['prompt_text'])):
  text = dataset.loc[i, 'prompt_text']
  answer = yandex(text)
  dataset.at[i, 'prompt_answer'] = answer
  if i % 20 == 0:
    print(i)
```

```python
for i in range(len(dataset['nominalization'])):
  text = dataset.loc[i, 'nominalization']
  if len(text) > 1:
    answer = yandex(text)
    dataset.at[i, 'nominalization_answer'] = answer
  if i % 20 == 0:
    print(i)
```

```python
for i in range(len(dataset['passivization'])):
  text = dataset.loc[i, 'passivization']
  if len(text) > 1:
    answer = yandex(text)
    dataset.at[i, 'passivization_answer'] = answer
  if i % 20 == 0:
    print(i)
```

```python
for i in range(len(dataset['scrambling'])):
  text = dataset.loc[i, 'scrambling']
  if len(text) > 1:
    answer = yandex(text)
    dataset.at[i, 'scrambling_answer'] = answer
  if i % 20 == 0:
    print(i)
```
